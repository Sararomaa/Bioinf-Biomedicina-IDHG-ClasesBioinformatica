#### 9.- Dada la lista _valores_, convierte cada número a string, pero con las siguientes reglas:
- Si es par: "par:<num>"
- Si es múltiplo de 4: "mult3:<num>"
- Si cumple ambas condiciones, concatenar: "par-mult3:<num>"
#### Devuelve la lista final.

```python
valores = [10,21,33,42,55]
```

```python
valores = [10, 21, 33, 42, 55]

resultado = []
for num in valores:
    es_par = (num % 2 == 0)
    es_mult3 = (num % 3 == 0)  # Evaluación de múltiplos de 3
    
    if es_par and es_mult3:
        resultado.append(f"par-mult3:{num}")
    elif es_par:
        resultado.append(f"par:{num}")
    elif es_mult3:
        resultado.append(f"mult3:{num}")
    else:
        resultado.append(str(num))

print(resultado)
```

#### 12.- Dado el diccionario _estudiantes_, crea un nuevo diccionario con la siguiente estructura:
- La clave debe ser el nombre en mayúsculas
- El valor debe ser otro diccionario con:
    - "promedio" como promedio de notas
    - "aprobado" como True si el promedio es >= 20, de lo contrario es False
    - "max_min" como tupla con (nota_max,nota_min)

```python
estudiantes = {
    "s1" : {"nombre":"ana","notas":[10,20,30]},
    "s2" : {"nombre":"luis","notas":[15,5]},
    "s3" : {"nombre":"marta","notas":[50,100,25]},
    "s4" : {"nombre":"pepe","notas":[40]}
}
```

```python
estudiantes = {
    "s1" : {"nombre":"ana","notas":[10,20,30]},
    "s2" : {"nombre":"luis","notas":[15,5]},
    "s3" : {"nombre":"marta","notas":[50,100,25]},
    "s4" : {"nombre":"pepe","notas":[40]}
}

# Diccionario donde guardaremos el resultado
resultado = {}

for datos in estudiantes.values():
    nombre_mayus = datos["nombre"].upper()
    notas = datos["notas"]
    
    promedio = sum(notas) / len(notas)
    aprobado = promedio >= 20
    max_min = (max(notas), min(notas))
    
    # Asignación correcta con '{' en la misma línea
    resultado[nombre_mayus] = {
        "promedio": promedio,
        "aprobado": aprobado,
        "max_min": max_min
    }

# Mostrar el resultado final formateado
import pprint
pprint.pprint(resultado)
```

#### 13.- Crea un archivo llamado _notas_alumnos.txt_ con este contenido:
#### Ana,10,20,30
#### Luis,5,15
#### Marta,100,50,25
#### Convierte cada línea en un diccionario con "nombre" capitalizado, "promedio" y "aprobado" si promedio >= 20.
#### Guarda la lista final en un archivo resultado.json

```python
import json

# crear el archivo notas_alumnos.txt con el contenido 
contenido = """Ana,10,20,30
Luis,5,15
Marta,100,50,25"""

with open("notas_alumnos.txt", "w", encoding="utf-8") as f:
    f.write(contenido)
#para leer el archivo y procesar cada linea
lista_alumnos = []

with open("notas_alumnos.txt", "r", encoding="utf-8") as f:
    for linea in f:
        linea = linea.strip()
        if not linea:
            continue
        
        partes = linea.split(",")
        nombre = partes[0].capitalize()
        notas = [float(n) for n in partes[1:]]
        
        promedio = sum(notas) / len(notas)
        aprobado = promedio >= 20
        
        alumno_dict = {
            "nombre": nombre,
            "promedio": promedio,
            "aprobado": aprobado
        }
        lista_alumnos.append(alumno_dict)

# Para guardar la lista final en resultado.json
with open("resultado.json", "w", encoding="utf-8") as f_json:
    json.dump(lista_alumnos, f_json, indent=4, ensure_ascii=False)

print("Contenido guardado en resultado.json:")
print(json.dumps(lista_alumnos, indent=4, ensure_ascii=False))
```

#### 14.- Crea el archivo _usuarios.json_con el contenido del diccionario _datos_. Abre el archivo de nuevo y convierte todos los nombres a mayúsculas, luego crea un archivo _usuarios.csv_ con formato:
- id,nombre,roles
- 1,ANA,user|admin
- 2,LUIS,user
- 3,MARTA,moderator|user

```python
import json
import csv

datos = {
    "1": {"nombre": "ana", "roles": ["user", "admin"]},
    "2": {"nombre": "luis", "roles": ["user"]},
    "3": {"nombre": "marta", "roles": ["moderator", "user"]}
}

with open("usuarios.json", "w", encoding="utf-8") as f_json:
    json.dump(datos, f_json, indent=4)

with open("usuarios.json", "r", encoding="utf-8") as f_json:
    datos_cargados = json.load(f_json)

#todos los nombres a mayúsculas
for user_info in datos_cargados.values():
    user_info["nombre"] = user_info["nombre"].upper()

with open("usuarios.csv", "w", newline="", encoding="utf-8") as f_csv:
    writer = csv.writer(f_csv)
    
    #Para el encabezado
    writer.writerow(["id", "nombre", "roles"])
    
    # Para cada fila de datos
    for user_id, user_info in datos_cargados.items():
        nombre = user_info["nombre"]
        roles_str = "|".join(user_info["roles"])
        writer.writerow([user_id, nombre, roles_str])

print("Contenido generado en usuarios.csv:\n")
with open("usuarios.csv", "r", encoding="utf-8") as f_csv:
    print(f_csv.read())
```

#### 15.- Con el archivo _fragmento-libro.txt_, ábrelo, cuenta la frecuencia de cada palabra (ignorando mayúsculas/minúsculas y signos). Devuelve un archivo llamado _frecuencias.json_ con las 5 palabras más comunes. 

```python
import json
import re
from collections import Counter

with open("fragmento-libro.txt", "r", encoding="utf-8") as f:
    texto = f.read()

palabras = re.findall(r'\b\w+\b', texto.lower())
conteo = Counter(palabras)
top_5 = dict(conteo.most_common(5))

with open("frecuencias.json", "w", encoding="utf-8") as f_json:
    json.dump(top_5, f_json, indent=4, ensure_ascii=False)

print("Top 5 palabras más comunes guardadas en frecuencias.json:")
print(json.dumps(top_5, indent=4, ensure_ascii=False))
```

#### 16.- Crea tres archivos CSV (_ventas1.csv_, _ventas2.csv_, _ventas3.csv_ con la misma información:
- producto,cantidad
- A,10
- B,5
#### Combínalos en un único archivo que se llema _total_ventas.csv_. Suma las cantidades por producto (si el producto se repite).


```python
import csv
from collections import defaultdict

contenido_csv = [
    ["producto", "cantidad"],
    ["A", 10],
    ["B", 5]
]

archivos_ventas = ["ventas1.csv", "ventas2.csv", "ventas3.csv"]
for nombre_archivo in archivos_ventas:
    with open(nombre_archivo, "w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerows(contenido_csv)
totales = defaultdict(int)

for nombre_archivo in archivos_ventas:
    with open(nombre_archivo, "r", encoding="utf-8") as f:
        reader = csv.reader(f)
        next(reader)  # Omitir el encabezado de cada archivo
        for producto, cantidad in reader:
            totales[producto] += int(cantidad)

with open("total_ventas.csv", "w", newline="", encoding="utf-8") as f_out:
    writer = csv.writer(f_out)
    writer.writerow(["producto", "cantidad"])  # Escribir el encabezado
    for producto, cantidad in totales.items():
        writer.writerow([producto, cantidad])

print("Contenido generado en total_ventas.csv:\n")
with open("total_ventas.csv", "r", encoding="utf-8") as f_check:
    print(f_check.read())
```

#### 17.- Crea un archivo _alumnos.json_ con el contenido siguiente:
- [{"nombre": "ana", "notas": [10, 20, 30]},{"nombre": "luis", "notas": [5, 15]},{"nombre": "marta", "notas": [100, 50, 25]}]
#### Lee el archivo y genera un diccionario donde la clave sea el nombre y el valor tenga el siguiente formato:
- {"promedio":X,"max":Y,"min":Z,"aprobado":True/False}


```python
import json

alumnos_datos = [
    {"nombre": "ana", "notas": [10, 20, 30]},
    {"nombre": "luis", "notas": [5, 15]},
    {"nombre": "marta", "notas": [100, 50, 25]}
]

with open("alumnos.json", "w", encoding="utf-8") as f:
    json.dump(alumnos_datos, f, indent=4)
with open("alumnos.json", "r", encoding="utf-8") as f:
    alumnos_cargados = json.load(f)

resultado = {}
for alumno in alumnos_cargados:
    nombre = alumno["nombre"]
    notas = alumno["notas"]
    
    promedio = sum(notas) / len(notas)
    nota_max = max(notas)
    nota_min = min(notas)
    aprobado = promedio >= 20
    
    resultado[nombre] = {
        "promedio": promedio,
        "max": nota_max,
        "min": nota_min,
        "aprobado": aprobado
    }

import pprint
pprint.pprint(resultado)
```
