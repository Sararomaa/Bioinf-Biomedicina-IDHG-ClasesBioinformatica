## import random

# ============================
#   CLASES PRINCIPALES
# ============================

class Virus:
    def __init__(self, load, replication_rate, name="Virus desconocido"):
        self.load = load
        self.replication_rate = replication_rate
        self.name = name

    def replicate(self):
        self.load *= self.replication_rate


class ImmuneSystem:
    def __init__(self, t_cells, nk_cells, macrophages, cytokines, tissue_damage):
        self.t_cells = t_cells
        self.nk_cells = nk_cells
        self.macrophages = macrophages
        self.cytokines = cytokines
        self.tissue_damage = tissue_damage

    def innate_response(self, virus):
        nk_effect = self.nk_cells * 0.05
        macro_effect = self.macrophages * 0.03
        total_reduction = nk_effect + macro_effect
        virus.load = max(0, virus.load - total_reduction)
        self.cytokines += (self.nk_cells + self.macrophages) * 0.02

    def adaptive_response(self, virus):
        t_effect = self.t_cells * 0.1
        virus.load = max(0, virus.load - t_effect)
        self.cytokines += self.t_cells * 0.03

    def apply_inflammation_effects(self):
        if self.cytokines > 80:
            self.tissue_damage += 12
        elif self.cytokines > 50:
            self.tissue_damage += 6
        elif self.cytokines > 30:
            self.tissue_damage += 3


# ============================
#   GRÁFICOS ASCII
# ============================

def ascii_bar(value, max_value=200, length=40):
    proportion = min(value / max_value, 1)
    filled = int(proportion * length)
    return "[" + "#" * filled + "-" * (length - filled) + "]"


# ============================
#   INTERFAZ Y ACCIONES
# ============================

def show_status(virus, immune, hour):
    print(f"\n=== Hora {hour} ===")
    print(f"Virus: {virus.name}")
    print(f"Carga viral: {virus.load:.1f} {ascii_bar(virus.load)}")
    print(f"Linfocitos T: {immune.t_cells:.1f}")
    print(f"Células NK: {immune.nk_cells:.1f}")
    print(f"Macrófagos: {immune.macrophages:.1f}")
    print(f"Citoquinas: {immune.cytokines:.1f}")
    print(f"Daño tisular: {immune.tissue_damage:.1f} {ascii_bar(immune.tissue_damage, max_value=100)}")


def ask_player_action():
    print("\nAcciones disponibles:")
    print("1) Activar linfocitos T")
    print("2) Potenciar células NK")
    print("3) Activar macrófagos")
    print("4) Aumentar interferones")
    print("5) Inducir apoptosis")
    print("6) Administrar antiviral")
    print("7) Descansar")
    choice = input("Elige una acción (1-7): ")
    return choice


def apply_action(choice, virus, immune):
    if choice == "1":
        immune.t_cells += 5
        immune.cytokines += 3
        print("Activas linfocitos T.")
    elif choice == "2":
        immune.nk_cells += 4
        immune.cytokines += 2
        print("Potencias células NK.")
    elif choice == "3":
        immune.macrophages += 4
        immune.cytokines += 2
        print("Activas macrófagos.")
    elif choice == "4":
        virus.replication_rate *= 0.8
        immune.cytokines += 6
        print("Aumentas interferones.")
    elif choice == "5":
        virus.load *= 0.7
        immune.tissue_damage += 10
        print("Induces apoptosis.")
    elif choice == "6":
        virus.replication_rate *= 0.6
        print("Administras antiviral.")
    elif choice == "7":
        immune.cytokines = max(0, immune.cytokines - 10)
        print("Descansas.")
    else:
        print("Acción no válida.")


def random_events(virus, immune):
    if random.random() < 0.1:
        virus.replication_rate *= 1.1
        print("EVENTO: El virus ha mutado.")
    if random.random() < 0.08:
        immune.t_cells = max(0, immune.t_cells - 3)
        print("EVENTO: Fatiga inmunológica.")
    if immune.cytokines > 90 and random.random() < 0.3:
        immune.tissue_damage += 15
        print("EVENTO: Tormenta de citoquinas.")


def game_over(virus, immune, hour):
    if virus.load <= 1:
        return "controlada"
    if immune.tissue_damage >= 100:
        return "fallo_organico"
    if hour >= 72:
        return "cronica"
    return None

def final_summary(result, virus, immune, hour):
    print("\n==============================")
    print("      RESUMEN DE LA PARTIDA")
    print("==============================")

        print("✔ La infección ha sido controlada con éxito.")

# ============================
#   PROGRAMA PRINCIPAL
# ============================

def main():
    show_instructions()

    while True:
        choice = main_menu()

        if choice == "1":
            difficulty = choose_difficulty()
            story_mode_intro()

            virus = Virus(
                load=difficulty["virus_load"],
                replication_rate=difficulty["replication_rate"],
                name=difficulty["virus_name"]
            )
            immune = ImmuneSystem(
                t_cells=difficulty["t_cells"],
                nk_cells=difficulty["nk_cells"],
                macrophages=difficulty["macrophages"],
                cytokines=difficulty["cytokines"],
                tissue_damage=0
            )

            hour = 0
            print("\nIniciando simulación...\n")

            while True:
                hour += 1
                show_status(virus, immune, hour)

                action = ask_player_action()
                apply_action(action, virus, immune)

                immune.innate_response(virus)
                virus.replicate()
                immune.adaptive_response(virus)
                immune.apply_inflammation_effects()
                random_events(virus, immune)

                result = game_over(virus, immune, hour)
                if result:
                    final_summary(result, virus, immune, hour)
                    break

        elif choice == "2":
            print("\nSaliendo del simulador...")
            break

        else:
            print("Opción no válida.\n")


if __name__ == "__main__":
    main()
